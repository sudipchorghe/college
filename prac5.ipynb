```python
def library_inventory_system():
    books = ()   # tuple to store book values

    while True:
        print("\n--- Library Inventory System ---")
        print("1. Enter book data")
        print("2. Display all book records")
        print("3. Calculate total number of books and average value")
        print("4. Display highest and lowest book values")
        print("5. Display premium books (>= given limit)")
        print("6. Search for a book value")
        print("7. Count occurrences of a book value")
        print("8. Remove duplicate records (convert to list & back)")
        print("9. Display tuple in reverse order")
        print("10. Find position (index) of a book value")
        print("0. Exit")

        # Handling potential input errors (like typing a letter)
        try:
            choice = int(input("Enter your choice: "))
        except ValueError:
            print("Please enter a valid number.")
            continue

        if choice == 1:
            n = int(input("Enter number of books to add: "))
            new_books = []
            for i in range(n):
                val = int(input(f"Enter book value {i+1}: "))
                new_books.append(val)
            books += tuple(new_books)
            print("Books added successfully!")

        elif choice == 2:
            print("Book Records:", books)

        elif choice == 3:
            if books:
                total = len(books)
                avg = sum(books) / total
                print("Total Books:", total)
                print("Average Value:", avg)
            else:
                print("No records available.")

        elif choice == 4:
            if books:
                print("Highest Value:", max(books))
                print("Lowest Value:", min(books))
            else:
                print("No records available.")

        elif choice == 5:
            limit = int(input("Enter premium limit: "))
            premium = tuple([b for b in books if b >= limit])
            print("Premium Books:", premium)

        elif choice == 6:
            val = int(input("Enter book value to search: "))
            if val in books:
                print("Book value exists in records.")
            else:
                print("Book value not found.")

        elif choice == 7:
            val = int(input("Enter book value to count: "))
            print(f"Book value {val} appears {books.count(val)} times.")

        elif choice == 8:
            books = tuple(set(books))
            print("Duplicates removed. Updated Records:", books)

        elif choice == 9:
            print("Reverse Order:", books[::-1])

        elif choice == 10:
            val = int(input("Enter book value to find position: "))
            if val in books:
                print(f"Position (Index): {books.index(val)}")
            else:
                print("Book value not found.")

        elif choice == 0:
            print("Exiting Library Inventory System...")
            break

        else:
            print("Invalid choice. Please try again.")

# Call the function to run the system
library_inventory_system()
```
