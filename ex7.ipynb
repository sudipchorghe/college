```python
def bank_account_system():
    # Dictionary to store {account_number: balance}
    accounts = {}

    while True:
        print("\n--- Bank Account Management System ---")
        print("1. Create Bank Accounts")
        print("2. Display All Account Details")
        print("3. Calculate Total & Average Balance")
        print("4. Find Highest & Lowest Balances")
        print("5. Display High-Value Accounts (Above Limit)")
        print("6. Search for an Account")
        print("7. Update Balance (Deposit/Withdrawal)")
        print("8. Remove Account (Close Account)")
        print("9. Display All Account Numbers & Balances Separately")
        print("10. Display Accounts Sorted by Account Number")
        print("0. Exit")

        try:
            choice = int(input("\nEnter your choice: "))
        except ValueError:
            print("Invalid input. Please enter a number.")
            continue

        if choice == 1:
            n = int(input("How many accounts to create? "))
            for _ in range(n):
                acc_num = input("Enter Account Number: ")
                bal = float(input(f"Enter Initial Balance for {acc_num}: "))
                accounts[acc_num] = bal
            print("Accounts created successfully!")

        elif choice == 2:
            print("\n--- Account Details ---")
            for acc, bal in accounts.items():
                print(f"Account: {acc} | Balance: ${bal:,.2f}")

        elif choice == 3:
            if accounts:
                total = sum(accounts.values())
                avg = total / len(accounts)
                print(f"Total Balance: ${total:,.2f}")
                print(f"Average Balance: ${avg:,.2f}")
            else:
                print("No accounts found.")

        elif choice == 4:
            if accounts:
                highest = max(accounts.values())
                lowest = min(accounts.values())
                print(f"Highest Balance: ${highest:,.2f}")
                print(f"Lowest Balance: ${lowest:,.2f}")
            else:
                print("No accounts found.")

        elif choice == 5:
            limit = float(input("Enter balance limit: "))
            high_val = {k: v for k, v in accounts.items() if v > limit}
            print(f"Accounts above ${limit}: {high_val}")

        elif choice == 6:
            acc_num = input("Enter account number to search: ")
            if acc_num in accounts:
                print(f"Account found! Current Balance: ${accounts[acc_num]:,.2f}")
            else:
                print("Account does not exist.")

        elif choice == 7:
            acc_num = input("Enter account number: ")
            if acc_num in accounts:
                action = input("Type 'D' for Deposit or 'W' for Withdrawal: ").upper()
                amount = float(input("Enter amount: "))
                if action == 'D':
                    accounts[acc_num] += amount
                    print("Deposit successful.")
                elif action == 'W':
                    if amount <= accounts[acc_num]:
                        accounts[acc_num] -= amount
                        print("Withdrawal successful.")
                    else:
                        print("Insufficient funds!")
            else:
                print("Account not found.")

        elif choice == 8:
            acc_num = input("Enter account number to close: ")
            if accounts.pop(acc_num, None):
                print(f"Account {acc_num} closed successfully.")
            else:
                print("Account not found.")

        elif choice == 9:
            print("Account Numbers:", list(accounts.keys()))
            print("Balances:", list(accounts.values()))

        elif choice == 10:
            print("\n--- Sorted Accounts (By Account Number) ---")
            sorted_acc = sorted(accounts.keys())
            for acc in sorted_acc:
                print(f"Account: {acc} | Balance: ${accounts[acc]:,.2f}")

        elif choice == 0:
            print("Exiting System. Goodbye!")
            break
        else:
            print("Invalid choice. Try again.")

# Run the program
bank_account_system()
```
