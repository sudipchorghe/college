```python
class ShoppingCart:
    def __init__(self, user_name):
        self.user_name = user_name
        self.cart = set()

    def add_product(self, product_id):
        self.cart.add(product_id)
        print(f"Product {product_id} added.")

    def remove_product(self, product_id):
        if product_id in self.cart:
            self.cart.remove(product_id)
            print(f"Product {product_id} removed.")
        else:
            print("Product ID not found.")

    def display_cart(self):
        print(f"\n--- {self.user_name}'s Shopping Cart ---")
        print(f"Items (Sorted): {sorted(list(self.cart))}")
        print(f"Total Unique Items: {len(self.cart)}")
        if self.cart:
            print(f"Highest ID: {max(self.cart)} | Lowest ID: {min(self.cart)}")

    def search_product(self, product_id):
        exists = product_id in self.cart
        print(f"Product {product_id} {'is' if exists else 'is NOT'} in the cart.")

    def show_premium(self, threshold):
        premium = {pid for pid in self.cart if pid >= threshold}
        print(f"Premium Products (ID >= {threshold}): {premium}")

def interactive_menu():
    name = input("Enter your name: ")
    user_cart = ShoppingCart(name)
    
    # Pre-populating a second cart for Union/Intersection demo
    other_cart = {101, 105, 500, 999}

    while True:
        print("\n1. Add Product\n2. Remove Product\n3. Display Cart\n4. Search Product\n5. Show Premium\n6. Compare with Guest Cart\n7. Exit")
        choice = input("Select an option (1-7): ")

        if choice == '1':
            pid = int(input("Enter Product ID to add: "))
            user_cart.add_product(pid)
        elif choice == '2':
            pid = int(input("Enter Product ID to remove: "))
            user_cart.remove_product(pid)
        elif choice == '3':
            user_cart.display_cart()
        elif choice == '4':
            pid = int(input("Enter Product ID to search: "))
            user_cart.search_product(pid)
        elif choice == '5':
            thresh = int(input("Enter Premium threshold ID: "))
            user_cart.show_premium(thresh)
        elif choice == '6':
            print(f"Guest Cart contains: {other_cart}")
            print(f"Combined (Union): {user_cart.cart | other_cart}")
            print(f"Common (Intersection): {user_cart.cart & other_cart}")
        elif choice == '7':
            print("Exiting...")
            break
        else:
            print("Invalid choice, try again.")

interactive_menu()
```
