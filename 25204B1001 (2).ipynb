5. Sum of digits of a number

```python
num = int(input("enter a number"))
sum = 0

while num > 0:
    sum += num % 10
    num //= 10
print ("sum of digits = " , sum)    
```

6. Reverse a number

```python
num = int(input("enter a number"))
rev = 0

while num > 0 :
    rev = rev * 10 + num % 10
    num //= 10
    print (rev)
```

7. Program to check if a number is a palindrome using for loop

```python
num = input("Enter a number: ")

if num == num[::-1]:   
    print(" number is an Palindrome")
else:
    print("number is Not Palindrome")
```

8. Keep taking input until the user enters 0

```python
while True:
    num = int(input("Enter a number (0 to stop): "))
    if num == 0:
        break
    print("You entered:", num)
```

9. calculator 

```python
while True:
    print("\n--- Calculator Menu ---")
    print("1. Addition , 2. Subtraction,3. Multiplication ")
    
    
    print("0. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 0:
        print("Exiting calculator...")
        break

    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if choice == 1:
        print("Result =", a + b)
    elif choice == 2:
        print("Result =", a - b)
    elif choice == 3:
        print("Result =", a * b)
    
    else:
        print("Invalid choice.")
```
